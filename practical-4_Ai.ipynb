```python
def manhattan(a, b):
    return abs(a[0]-b[0]) + abs(a[1]-b[1])

def astar(grid, start, goal):
    open_list = [(start, [start], 0)]  # (node, path, cost)
    visited = set()

    while open_list:
        open_list.sort(key=lambda x: x[2] + manhattan(x[0], goal))  # f = g + h
        node, path, cost = open_list.pop(0)
        if node == goal:
            return path, cost
        if node in visited:
            continue
        visited.add(node)

        x, y = node
        for dx, dy in [(1,0), (-1,0), (0,1), (0,-1)]:
            nx, ny = x+dx, y+dy
            if 0 <= nx < len(grid) and 0 <= ny < len(grid[0]) and grid[nx][ny] != 'X':
                new_cost = cost + grid[nx][ny]
                open_list.append(((nx, ny), path + [(nx, ny)], new_cost))
    return None, float('inf')

grid = [
    [1, 1, 5],
    [1, 'X', 1],
    [1, 1, 1]
]

path, cost = astar(grid, (0,0), (2,2))
print("Path:", path)
print("Cost:", cost)
```
